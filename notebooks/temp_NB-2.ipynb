```python
import sys, json, torch, time
sys.path.insert(0, "/workspace/shared/audit_validator")

from src.validator_agent import AuditAgent
from src.utils import log_gpu_memory

log_gpu_memory("Before audit")

agent  = AuditAgent()
t_start = time.time()
result = agent.run(
    "/workspace/shared/audit_validator/data/sample_docs/demo_enterprise_contract.pdf",
    max_chunks=12
)
total_time = round(time.time() - t_start, 2)

log_gpu_memory("After audit")

score = result["score_data"]
metrics = result["metrics"]

slide4_metrics = {
    "model_llm"         : "Qwen/Qwen2.5-7B-Instruct",
    "model_embed"       : "BAAI/bge-small-en-v1.5",
    "gpu"               : "AMD Instinct MI300X — 206GB",
    "rocm_version"      : "7.2",
    "total_rules"       : 22,
    "frameworks"        : ["GDPR","SOX","HIPAA","PCI_DSS","Insurance_Compliance"],
    "chunks_processed"  : metrics["total_chunks"],
    "rules_checked"     : metrics["total_rules_checked"],
    "total_tokens"      : metrics["total_tokens"],
    "avg_latency_sec"   : metrics["avg_latency_sec"],
    "total_time_sec"    : total_time,
    "overall_score"     : score["overall_score"],
    "risk_level"        : score["risk_level"],
    "framework_scores"  : score["framework_scores"],
    "status_counts"     : score["status_counts"],
    "gpu_allocated_gb"  : round(torch.cuda.memory_allocated()/1e9, 2),
    "gpu_reserved_gb"   : round(torch.cuda.memory_reserved()/1e9, 2),
    "gpu_total_gb"      : round(torch.cuda.get_device_properties(0).total_memory/1e9, 2)
}

with open("/workspace/shared/audit_validator/logs/slide4_final_metrics.json", "w") as f:
    json.dump(slide4_metrics, f, indent=2)

print("\n" + "="*60)
print("📊 SLIDE 4 FINAL METRICS")
print("="*60)
print(json.dumps(slide4_metrics, indent=2))
print("\n✅ Saved to logs/slide4_final_metrics.json")
```
